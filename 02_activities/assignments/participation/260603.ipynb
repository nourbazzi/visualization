Three important qualities of data visualization: aesthetic, substantive (visualization accurately presents data), and perceptual (can understand the message)

effective visualization depends on purpose, audience, and medium.

```python
Gestalt principles are a set of cognitive theories forhow people tend to organize visual information. 
Things to consider: proximity, similarity, continuity, closure, enclosure, connection
```

```python
Things that affect cognitive load: 
- familiar vs rare chart types
- accurate vs approximate interpretation
- concise vs detailed composition
- explanatory vs exploratory composition
```

Four conventions of data visualization accordin to Kennedy et al. (2016):
- 2D image
- clean layout
- geometric shapes/lines
- inclusion of data sources

Provenance rhetoric:the idea that the inclusion of a data source with our graphic signals transparency to the audience
